```python
%sql
CREATE OR REPLACE VIEW IDENTIFIER(:catalog || '.gold.customers_secure_view') AS 
SELECT
customer_id,
name,
CONCAT('*****@', SPLIT(email, '@')[1]) AS email,
city,
state,
CONCAT('********', RIGHT(phone, 4)) AS phone
FROM IDENTIFIER(:catalog ||'.silver.customers_scd_1');
```
