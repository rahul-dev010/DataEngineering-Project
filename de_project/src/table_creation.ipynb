```python
dbutils.widgets.text("catalog","default")

catalog=dbutils.widgets.get("catalog")
print(catalog)
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER('${catalog}.bronze.customers')
(
    customer_id INT,
    name STRING,
    email STRING,
    city STRING,
    state STRING,
    signup_date DATE,
    phone STRING,
    customer_segment STRING,

    -- Metadata Columns
    ingestion_timestamp TIMESTAMP,
    file_name STRING,
    file_path STRING
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER('${catalog}.bronze.products')
(
    product_id INT,
    product_name STRING,
    category STRING,
    price DECIMAL(10,2),
    supplier_id INT,
    sku STRING,

    ingestion_timestamp TIMESTAMP,
    source_file_name STRING,
    file_path STRING
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER('${catalog}.bronze.suppliers')
(
    supplier_id INT,
    supplier_name STRING,
    contact_email STRING,
    country STRING,

    ingestion_timestamp TIMESTAMP,
    source_file_name STRING,
    file_path STRING
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER('${catalog}.bronze.sales')
(
    sale_id INT,
    customer_id INT,
    product_id INT,
    quantity INT,
    sale_amount DECIMAL(12,2),
    discount DECIMAL(5,2),
    sale_date DATE,
    region STRING,
    payment_method STRING,
    order_status STRING,

    ingestion_timestamp TIMESTAMP,
    source_file_name STRING,
    file_path STRING
)
USING DELTA;
```
