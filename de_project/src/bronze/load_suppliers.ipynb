```python
%sql

COPY INTO IDENTIFIER('${catalog}.bronze.suppliers')
FROM (
    SELECT
        supplier_id,
        supplier_name,
        contact_email,
        country,
        current_timestamp() AS ingestion_timestamp,
        _metadata.file_name AS file_name,
        _metadata.file_path AS file_path
    FROM '/Volumes/${catalog}/bronze/raw/suppliers'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
    'header' = 'true',
    'inferSchema' = 'true'
)
COPY_OPTIONS (
    'mergeSchema' = 'true'
);
```
