```python
%sql
COPY INTO IDENTIFIER ('${catalog}.bronze.products')
FROM (
    SELECT
        product_id,
        product_name,
        category,
        CAST(price AS DECIMAL(10,2)) AS price,
        supplier_id,
        sku,
        current_timestamp() AS ingestion_timestamp,
        _metadata.file_name AS file_name,
        _metadata.file_path AS file_path
    FROM '/Volumes/${catalog}/bronze/raw/products'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
    'header' = 'true',
    'inferSchema' = 'true'
)
COPY_OPTIONS (
    'mergeSchema' = 'true'
);
```
