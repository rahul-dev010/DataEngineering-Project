```python
%sql

COPY INTO IDENTIFIER('${catalog}.bronze.sales')
FROM (
    SELECT
        sale_id,
        customer_id,
        product_id,
        CAST(quantity AS INT) AS quantity,
        CAST(sale_amount AS DECIMAL(12,2)) AS sale_amount,
        CAST(discount AS DECIMAL(5,2)) AS discount,
        sale_date,
        region,
        payment_method,
        order_status,
        current_timestamp() AS ingestion_timestamp,
        _metadata.file_name AS file_name,
        _metadata.file_path AS file_path
    FROM '/Volumes/${catalog}/bronze/raw/sales'
)
FILEFORMAT = CSV
FORMAT_OPTIONS (
    'header' = 'true',
    'inferSchema' = 'true'
)
COPY_OPTIONS (
    'mergeSchema' = 'true'
);
```
